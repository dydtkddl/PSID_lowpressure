```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score

# Seaborn 스타일
sns.set_theme(style="ticks", context="talk")

# 데이터 불러오기
DFQT = pd.read_csv("../try13/Ar_293_0.01_to_5__struct+input__qt_then_rd__cat_TRAIN_RATIO0.8_QTFRAC_0.8/trial_001/predictions_holdout_trial001.csv")
DFST = pd.read_csv("../try13/Ar_293_0.01_to_5__struct__random_struct__cat_TRAIN_RATIO0.8_QTFRAC_0.8//trial_001/predictions_holdout_trial001.csv")
DFRD = pd.read_csv("../try13/Ar_293_0.01_to_5__struct+input__random_with_input__cat_TRAIN_RATIO0.8_QTFRAC_0.8/trial_001/predictions_holdout_trial001.csv")

# 색상 매핑
color_map = {
    "QT": "#DD8452",   # 주황
    "ST": "#7A7A7A",   # 회색
    "RD": "#4C72B0"    # 파랑
}

# R² 계산
r2_st = r2_score(DFST["y_true"], DFST["y_pred"])
r2_rd = r2_score(DFRD["y_true"], DFRD["y_pred"])
r2_qt = r2_score(DFQT["y_true"], DFQT["y_pred"])

plt.figure(figsize=(7,7))

# ST
plt.scatter(DFST["y_true"], DFST["y_pred"], 
            color=color_map["ST"], alpha=0.6, s=80, marker="o",
            label=f"ST (R²={r2_st:.3f})")

# RD
plt.scatter(DFRD["y_true"], DFRD["y_pred"], 
            color=color_map["RD"], alpha=0.6, s=60, marker="s",
            label=f"RD (R²={r2_rd:.3f})")

# QT
plt.scatter(DFQT["y_true"], DFQT["y_pred"], 
            color=color_map["QT"], alpha=0.7, s=40, marker="^",
            label=f"QT (R²={r2_qt:.3f})")

# y=x 대각선
lims = [
    min(DFQT["y_true"].min(), DFST["y_true"].min(), DFRD["y_true"].min(),
        DFQT["y_pred"].min(), DFST["y_pred"].min(), DFRD["y_pred"].min()),
    max(DFQT["y_true"].max(), DFST["y_true"].max(), DFRD["y_true"].max(),
        DFQT["y_pred"].max(), DFST["y_pred"].max(), DFRD["y_pred"].max())
]
plt.plot(lims, lims, "k--", lw=2, label="y = x")

# 스타일
plt.xlabel("True values")
plt.ylabel("Predicted values")
plt.title("Parity Plot (Ar, 293 K)")
plt.legend()
sns.despine()
plt.tight_layout()
plt.show()
```
